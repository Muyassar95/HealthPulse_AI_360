## 1. Importation des bibliothèques

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Créer un premier tableau fictif

```python

data = {
    "departement": ["Paris", "Nord", "Rhône", "Gironde", "Yvelines"],
    "patients_diabete": [25000, 42000, 21000, 18000, 16000],
    "population": [500000, 700000, 400000, 350000, 300000]
}

df = pd.DataFrame(data)

df
```

## 3. Explorer le DataFrame

```python

df.head()
```

```python

df.info()
```

```python

df.describe()
```

## 4. Calculer la prévalence (%)

```python

df["prevalence_pct"] = (
    df["patients_diabete"] / df["population"]
) * 100

df
```

## 5. Créer le premier graphique

```python

plt.figure(figsize=(8, 4))

plt.bar(df["departement"], df["prevalence_pct"])

plt.title("Prévalence fictive du diabète")
plt.xlabel("Département")
plt.ylabel("Prévalence (%)")

plt.show()
```

## 6. Filtrage des données

```python
df[df["prevalence_pct"] > 5.3]
```

```python
df.to_csv("data/processed/diabete_exercice.csv", index=False)
```
